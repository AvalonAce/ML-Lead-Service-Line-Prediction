```python
import os
import pandas as pd
import requests
import json
```

```python
with open('raw/home_jsons.json') as file:
    homes = json.load(file)

zipcode_info = set()
for home in homes:
    #print(home)
    zip = None
    try:
        zip = home['property'][0]['location']['geoIdV4']['ZI']
        zipcode_info.add(zip)
    except KeyError as e:
        print(e)

print(len(zipcode_info))
```

```python
from dotenv import load_dotenv

load_dotenv('../../.env')
ATTOM_API_KEY = os.getenv('ATTOM_API_KEY')

def get_attom_zipcode_data(zipID):
    url = f"https://api.gateway.attomdata.com/v4/neighborhood/community?geoIdv4={zipID}"
    headers = {
        "apikey": ATTOM_API_KEY
    }
    response = None
    try:
        response = requests.get(url, headers=headers, timeout=5)
    except requests.exceptions.RequestException as e:
        print(f"request failed {e}")
        return False, None
    
    if response.status_code == 200:
        data = response.json()
        status = data.get('status', {})
        if status.get('msg') == 'SuccessWithResult':
            return True, data
        else:
            return False, data
    else:
        return False, response.json()

# jsonarr = []
# for zip in zipcode_info:
#     ok, data = get_attom_zipcode_data(zip)
#     if ok:
#         jsonarr.append(data)
#     print(jsonarr)
```
